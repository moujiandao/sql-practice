```python
# Use this when you're working with entire columns (not filtering rows):

# SELECT a column
df['country']  # returns a Series (single column)
df[['country', 'price']]  # returns a DataFrame (multiple columns)

# CREATE or UPDATE a column
df['new_column'] = some_values  # adds/overwrites entire column
df['price_doubled'] = df['price'] * 2  # column math

# df.loc[] - Row Filtering + Optional Column Selection
# Use this when you need to filter rows based on conditions:
df.loc[df['price'] > 100]  # filtering ROWS
df.loc[df['country'] == 'Italy', 'price']  # filter rows, select specific column

# When to use: Filtering, conditional selection, row-based operations.
# Simple Rule:
# Columns only? → df['column']
# Filtering rows? → df.loc[condition]
```

```python
df.loc[df['price'].notnull()]  # WHERE price IS NOT NULL
df.loc[df['description'].isnull()]  # WHERE description IS NULL
```

```python
# AND - both conditions must be true
df.loc[(df['country'] == 'Italy') & (df['price'] > 50)]

# OR - either condition works
df.loc[(df['country'] == 'Italy') | (df['points'] >= 90)]

# Instead of: (df['country'] == 'Italy') | (df['country'] == 'France') | ...
df.loc[df['country'].isin(['Italy', 'France', 'Spain'])]  # SQL's IN clause


# Less readable
df.loc[(df['country'] == 'Italy') & (df['price'].notnull()) & (df['points'] >= 90)]

# Better - break it down if complex
italy_wines = df.loc[df['country'] == 'Italy']
italy_wines = italy_wines.loc[italy_wines['price'].notnull()]
high_rated = italy_wines.loc[italy_wines['points'] >= 90]


# Quick Mental Model:
# Think of .loc[] as: "Give me [rows based on this condition], [and these specific columns if I want]"
df.loc[CONDITION]  # all columns
df.loc[CONDITION, 'column_name']  # specific column
df.loc[CONDITION, ['col1', 'col2']]  # multiple columns
```

```python
# Single cell - rarely used but good to know
df.loc[0, 'country']  # row with index 0, column 'country'

# Entire row(s)
df.loc[0]  # all columns for index 0
df.loc[[0, 5, 10]]  # specific rows by index

# All rows matching a condition (this is 80% of what you'll do)
df.loc[df['country'] == 'Italy']  # WHERE country = 'Italy' in SQL
```
